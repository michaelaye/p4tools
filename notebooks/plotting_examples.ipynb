# Plotting examples
> plotting stuff with `p4tools`

```python
from p4tools import io
from p4tools import markings as m
from p4tools import plotting
```

Defining three tile_ids for later use:

```python
tid1, tid2, tid3 = "APF00006tp APF0000tld APF000018t".split()
```

## Plot original tile
Plot the HiRISE tile used as input for the Citizen scientist:

```python
# | eval: false
plotting.plot_original_tile(tid1)
```

## Plot original and fans
Plot the HiRISE tile without and with the fans for a given `tile_id`:

```python
# | eval: false
plotting.plot_original_and_fans(tid2)
```

## Plot blotches for tile
Simply plot all blotches for a given `tile_id` on top of the HiRISE tile:

```python
plotting.plot_blotches_for_tile(tid3)
```

## Get fans for tile
get a dataframe with all fans for a given `tile_id`:

```python
fans = io.get_fans_for_tile(tid2)
```

```python
fans.head()
```

## Plot longest fan of a tile
use `distance` to find the longest fan in this dataframe:

```python
fans.columns
```

```python
maxrow = fans.loc[fans.distance.idxmax()]
```

```python
maxrow
```

use the `Fans` (or `Blotch`) objects of the `marking` module (here abbreviated to `m` during import above):

```python
fan = m.Fan(fans.iloc[0])
fan
```

and then let the fan object plot itself:

```python
fan.plot()
```

or, as above, use the plotting module function to plot all of the tile:

```python
plotting.plot_fans_for_tile(tid2)
```
